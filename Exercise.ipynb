```python
# Setup
target_name = "PASCUAL"
SEED_NUM = 1
redacted_name = ""

# Logic
vowels = "AEIOU"
for char in target_name:
    if char in vowels:
        redacted_name += str(SEED_NUM)
    else:
        redacted_name += char

print("Original Surname:", target_name)
print("SEED_NUM Used:", SEED_NUM)
print("Redacted Output:", redacted_name)
```

```python
# Setup
engineer_name = "LLOYD"   # Unang letra ay 'L' (ASCII 76)
SEED_NUM = 1
loops = 4                 # Batay sa Total Loops (C) mo na 4

# Logic
first_letter_ascii = ord(engineer_name[0])  # 'L' -> 76
verification_hash = (first_letter_ascii * SEED_NUM) - loops

print("First letter ASCII (A):", first_letter_ascii)
print("SEED_NUM (B):", SEED_NUM)
print("Total Loops (C):", loops)
print("Execution Log:", f"{first_letter_ascii}x{SEED_NUM}-{loops}")
print("Verification hash:", verification_hash)
```
